```python
# Data Preprocessing
```

```python
df['PURPOSE'].fillna("not",inplace=True)
```

```python
df.head()
```

```python
df['START_DATE']=pd.to_datetime(df['START_DATE'], errors='coerce')
```

```python
df['END_DATE']=pd.to_datetime(df['END_DATE'], errors='coerce')
```

```python
df.info()
```

```python
from datetime import datetime

df["Date"]=pd.DatetimeIndex(df['START_DATE']).date
df["Time"]=pd.DatetimeIndex(df['START_DATE']).hour
```

```python
df.head()
```

```python
df['Day-Night']=pd.cut(x=df['Time'], bins=[0,10,15,19,24],labels=['Morning','Afternoon','Evening','Night'])
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.info()
```

```python
df.shape
```

DATA Visualization

```python
plt.figure(figsize=(20,8))
plt.subplot(1,2,1)
sns.countplot(x='CATEGORY',data=df)
plt.show()
```

```python
plt.figure(figsize=(20,5))
plt.subplot(1,2,1)
sns.countplot(x='CATEGORY',data=df)

plt.figure(figsize=(25,5))
plt.subplot(1,2,2)
sns.countplot(x='PURPOSE',data=df)
plt.xticks
plt.show()
```

```python
df['Day-Night'] = df['Day-Night'].cat.add_categories("Unknown")
df['Day-Night'] = df['Day-Night'].fillna("Unknown")
```

```python
sns.countplot(x='Day-Night',data=df)
plt.show()
```

```python
df.info()
```

```python
df['Month']=df['START_DATE'].dt.month
month_label={1:'Jan',2:'Feb',3:'Mar',4:'Apr',5:'May',6:'Jun',7:'Jul',8:'Aug',9:'Sep',10:'Oct',11:'Nov',12:'Dec'}
df['Month']=df.Month.map(month_label)
mon=df.Month.value_counts(sort=False)
```

```python
print(df.columns.tolist())
```

```python
dataset=pd.DataFrame({
    'Month':mon.values,
    'Value_count': df.groupby('Month',sort=False)['MILES'].max()
})
p=sns.lineplot(data=dataset)
p.set(xlabel='Month',ylabel='Miles')
plt.show()
```

```python
print(df.columns.tolist())
```

```python
df['Week']=df.START_DATE.dt.weekday
week_label={1:'Mon',2:'Tue',3:'Wed',4:'Thu',5:'Fri',6:'Sat',7:'Sun'}
df['Week']=df.Week.map(week_label)
```

```python
week_label=df.Week.value_counts()
sns.barplot(x=week_label.index,y=week_label)
plt.xlabel('DAY')
plt.ylabel('Value_count')
plt.show()
```
