```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x1=[1,2,3,4,5]
y1=[1,2,3,4,5]
plt.plot(x1,y1)
plt.show()
sns.lineplot(x=x1,y=y1)
plt.show()
```

```python
x1=[1,2,3,4,5]
y1=[1,2,3,4,5]
df=pd.DataFrame({"x":x1,'y':y1})
df
sns.lineplot(x="x",y="y",data=df)
plt.show()
```
