Array Initialization Method

````python
#Zeros Array
zero_arr=np.zeros((2,5))
print(zero_arr)
````

````python
#One Array
one_arr=np.ones((2,3))
print(one_arr)
````

````python
#Full Array
full_arr=np.full((3,7),8)
print(full_arr)
````

````python
#Identity Matrix
id_arr=np.eye(6)
print(id_arr)
````

````python
print(np.empty(2))
````

````python
print(np.arange(1,6,1))
````

````python
#Random values array
r_arr=np.random.randint(1,67,(3,6))
print(r_arr)
````

**Slicing in array**

````python
arr1=np.array([1,2,3,45,6,7,8,9])
print(arr1[2:5])
print(arr1[::1])
````

````python
arr2=np.array([[1,2,9],[5,6,0]])
print(arr2[1])
print(arr2[1:])
print(arr2[1][0:])
print(arr2[:1])
````

#Array reshaping and flattening

````python
Arr1=np.array([1,2,3,4,5,6])
reshape = Arr1.reshape(3,2)
print(reshape)
````

````python
flat=reshape.flatten()
print(flat)
````

**Stacking and Splitting**

````python
a=np.array([1,2,3,4,5])
b=np.array([6,7,8,9,0])
print(np.vstack((a,b)))
print(np.hstack((a,b)))
````
