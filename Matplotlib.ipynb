```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [12000, 15000, 14000, 18000, 22000]

plt.plot(months, sales)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
sales_2024 = [12,15,14,18,22]
sales_2025 = [14,16,18,21,25]

plt.plot(months, sales_2024, label="2024")
plt.plot(months, sales_2025, label="2025")

plt.legend()
plt.show()
```

```python
products = ["A", "B", "C", "D"]
revenue = [50, 80, 30, 60]

plt.bar(products, revenue)
plt.title("Product Revenue")
plt.show()
```

```python
share = [40, 30, 20, 10]
labels = ["India", "USA", "UK", "Other"]

plt.pie(share, labels=labels, autopct="%1.1f%%")
plt.title("Market Share")
plt.show()
```
