```python
a={
    "ID":1,
    "NAME":"Anish Kumar",
    "AGE":20

}
print(a)
a['COUNTRY']="India"
print(a.keys())
print(a.values())
```

```python
a={
    "ID":1,
    "NAME":"Anish Kumar",
    "AGE":20,
    "COUNTRY":"India"

}
print(a)
print(a.pop("ID"))
print(a.keys())
print(a.values())
    
```
