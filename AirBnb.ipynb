```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/Airbnb_Open_Data.csv')
df
```

```python
df.info()
```

```python
df.columns
```

# DATA CLEANING

```python
df.isnull().sum()
```

# Handling Missing Values

```python
df['last review']=pd.to_datetime(df['last review'] , errors='coerce')
```

```python
df.info()
```

```python
df.fillna({'reviews per month':0, 'last review':df['last review'].min()},inplace=True)
df.head()
```

```python
df.dropna(subset=['NAME','host name'],inplace=True)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df=df.drop(columns=['license','house_rules'],errors='ignore')
df
```

```python
df['price']=df['price'].replace('[\$,]',"",regex=True).astype(float)
df['service fee']=df['service fee'].replace('[\$,]',"",regex=True).astype(float)
```

```python
df.head()
```

# Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

```python
df.describe()
```

# Visualisation

```python
plt.figure(figsize=(10,6))
sns.histplot(df['price'],kde=True,bins=50,color='yellow')
plt.show()
```

```python
df['room type']
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x='room type',data=df)
plt.title('Room Types')
plt.xlabel('Room')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(20,6))
sns.countplot(x='neighbourhood group',order=df['neighbourhood group'].value_counts().index,data=df)
plt.title('Neighbourhood distribution')
plt.xlabel('Neighbourhood')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(20,6))
sns.boxplot(x='room type',y='price',hue='room type',data=df,palette='Set1')
plt.title('Price Distribution by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Price')
plt.show()
```

```python
df['last review'] = pd.to_datetime(df['last review'])

reviews_over_time = df.groupby(
    df['last review'].dt.to_period('M')
).size()

plt.figure(figsize=(12, 6))

reviews_over_time.plot(
    kind='line',
    color='red'
)

plt.title('Number of Reviews Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Reviews')

plt.show()
```
