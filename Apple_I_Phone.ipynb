**Iphone Analysis**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

```python
df=pd.read_csv("/content/apple_products.csv")
df
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.drop_duplicates
```

```python
high_rating=df.sort_values(by=['Star Rating'],ascending=False)
```

```python
high_rating.head(10)
print(high_rating['Product Name'])
```

```python
iphones=high_rating['Product Name'].value_counts()
label=iphones.index
counts=high_rating['Number Of Ratings']
fig=px.bar(high_rating,x=label,y=counts,title='Iphone Rating')
fig.show()
```

```python
iphones=high_rating['Product Name'].value_counts()
label=iphones.index
counts=high_rating['Number Of Reviews']
fig=px.bar(high_rating,x=label,y=counts,title='Iphone Reviews')
fig.show()
```

```python
fig=px.scatter(data_frame=df,x='Number Of Ratings',y='Sale Price',title='Number of sales on the basis of rating',size='Discount Percentage',trendline='ols')
fig.show()
```

```python
figure = px.scatter(data_frame=df,x='Number Of Ratings',y='Discount Percentage',size='Sale Price',title='Number of sales on the basis of discount',trendline='ols')
figure.show()
```
